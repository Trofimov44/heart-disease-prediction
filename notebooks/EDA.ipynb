4.4 Статистические тесты

```python
df.info()
```

```python
#Для числовых признаков — тест Манна-Уитни (непараметрический, не требует нормальности)
from scipy.stats import mannwhitneyu

# Абстрактно:
group_healthy = df[df['condition'] == 0]['thalach']
group_sick = df[df['condition'] == 1]['thalach']
stat, p_value = mannwhitneyu(group_healthy, group_sick)
print(stat)
print(p_value)
```

```python
from scipy.stats import mannwhitneyu

group_healthy = df[df['condition'] == 0]['oldpeak']
group_sick = df[df['condition'] == 1]['oldpeak']
stat, p_value = mannwhitneyu(group_healthy, group_sick)
print(stat)
print(p_value)
```

```python
#Для категориальных признаков — тест хи-квадрат:
from scipy.stats import chi2_contingency

table = pd.crosstab(df['cp'], df['condition'])
chi2, p, dof, expected = chi2_contingency(table)
print(chi2,'\n', p,'\n', dof,'\n', expected)
```

4.5 Мультиколлинеарность

Корреляционная матрица показывает пары. Но бывает, что один признак линейно выражается через комбинацию нескольких других. Это ловит VIF (Variance Inflation Factor).

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools import add_constant

X = df[['age', 'trestbps', 'chol_log', 'thalach', 'oldpeak']]
X_with_const = add_constant(X)  # добавляет столбец единиц

vif_data = pd.DataFrame({
    'feature': X.columns,
    'VIF': [variance_inflation_factor(X_with_const.values, i+1) 
            for i in range(X.shape[1])]
})
print(vif_data)
```
